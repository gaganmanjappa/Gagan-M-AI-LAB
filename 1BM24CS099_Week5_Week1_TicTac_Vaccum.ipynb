```python
print("GAGAN M")
print("WEEK-1 AI LAB")
print("TIC-TAC-TOE GAME")

# Set up the game board as a list
board = ["-", "-", "-",
         "-", "-", "-",
         "-", "-", "-"]


# Define a function to print the game board
def print_board():
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("---------")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("---------")
    print(board[6] + " | " + board[7] + " | " + board[8])


# Define a function to handle player's turn
def take_turn(player):
    print(player + "'s turn")

    position = input("Choose a position from 1-9: ")

    # Check whether input is valid
    while position not in ["1", "2", "3", "4", "5", "6", "7", "8", "9"]:
        position = input("Invalid input. Choose a position from 1-9: ")

    position = int(position) - 1

    # Check whether the position is already occupied
    while board[position] != "-":
        position = int(input("Position already taken. Choose another position: ")) - 1

    board[position] = player

    print_board()


# Define a function to check the winner
def check_winner(player):

    # Check rows
    if board[0] == player and board[1] == player and board[2] == player:
        return True

    if board[3] == player and board[4] == player and board[5] == player:
        return True

    if board[6] == player and board[7] == player and board[8] == player:
        return True

    # Check columns
    if board[0] == player and board[3] == player and board[6] == player:
        return True

    if board[1] == player and board[4] == player and board[7] == player:
        return True

    if board[2] == player and board[5] == player and board[8] == player:
        return True

    # Check diagonals
    if board[0] == player and board[4] == player and board[8] == player:
        return True

    if board[2] == player and board[4] == player and board[6] == player:
        return True

    return False


# Define a function to check for draw
def check_draw():
    return "-" not in board


# Main game
print_board()

player = "X"

while True:

    take_turn(player)

    if check_winner(player):
        print(player + " wins!")
        break

    if check_draw():
        print("It's a draw!")
        break

    # Change player
    if player == "X":
        player = "O"
    else:
        player = "X"
```

```python
print("GAGAN M")
print("WEEK-1 AI LAB")
print("VACUUM CLEANER AGENT")
print("SIMPLE REFLEX AGENT")


# Set up the environment
room = {
    "A": "Dirty",
    "B": "Dirty"
}

# Starting position of vacuum cleaner
position = "A"


# Define the simple reflex agent
def simple_reflex_agent(location, status):

    if status == "Dirty":
        return "Suck"

    elif location == "A":
        return "Move Right"

    else:
        return "Move Left"


# Run the agent
for i in range(4):

    print("\nCurrent Location:", position)
    print("Room Status:", room[position])

    action = simple_reflex_agent(position, room[position])

    print("Action:", action)

    if action == "Suck":
        room[position] = "Clean"

    elif action == "Move Right":
        position = "B"

    elif action == "Move Left":
        position = "A"


print("\nFinal Environment:")
print(room)
```
