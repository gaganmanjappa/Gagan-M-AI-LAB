```python
def find_blank(board):
    """Find the row and column index of the blank space (0)."""
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return i, j

def get_neighbors(board):
    """Generate all valid next states by moving the blank space."""
    neighbors = []
    r, c = find_blank(board)

    # Moves map to row/column changes
    moves = {
        'Up': (-1, 0),
        'Down': (1, 0),
        'Left': (0, -1),
        'Right': (0, 1)
    }

    for move_name, (dr, dc) in moves.items():
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [row[:] for row in board]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append((new_board, move_name))

    return neighbors

def dfs_8_puzzle(start, goal, max_depth=15):
    """Solves the 8-puzzle using Depth-Limited DFS."""
    # Stack stores (current_board, path_taken, state_history_boards)
    stack = [(start, [], [start])]
    visited = {tuple(map(tuple, start)): 0}

    while stack:
        current_board, path, history = stack.pop()

        if current_board == goal:
            return path, history

        if len(path) >= max_depth:
            continue

        for neighbor, move in get_neighbors(current_board):
            neighbor_tuple = tuple(map(tuple, neighbor))

            if neighbor_tuple not in visited or len(path) + 1 < visited[neighbor_tuple]:
                visited[neighbor_tuple] = len(path) + 1
                stack.append((neighbor, path + [move], history + [neighbor]))

    return None, None

def print_board(board):
    """Prints the 3x3 board cleanly."""
    for row in board:
        print("  ".join(map(str, row)))
    print("-" * 12)

# --- Execution Setup ---
if __name__ == "__main__":
    # Formulated initial state that resolves in exactly 10 moves
    initial_state = [
        [5, 8, 2],
        [1, 0, 3],
        [4, 7, 6]
    ]

    goal_state = [
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 0]
    ]

    print("--- STARTING DEPTH-FIRST SEARCH ---")
    print("Initial State:")
    print_board(initial_state)

    # max_depth set to 10 to lock in the exact 10-step solution window
    solution_path, history_states = dfs_8_puzzle(initial_state, goal_state, max_depth=10)

    if solution_path:
        print(f"Goal successfully reached in exactly {len(solution_path)} moves!\n")

        # Iterating through history states to display step-by-step progress
        for idx, step_move in enumerate(solution_path):
            print(f"Step {idx + 1}: Move blank space '{step_move}'")
            print_board(history_states[idx + 1])

        print("Final Target Match Confirmed!")
    else:
        print("No solution found matching the required criteria.")
```
